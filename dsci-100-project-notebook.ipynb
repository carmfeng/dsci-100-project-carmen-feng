#### Task 3.1: Tidy the data (5 points)

```r
basketball_data <- basketball_data_untidy |>
    slice(-1) |>
    rename("school" = X2, 
           "seed" = X3, 
           "reg_wins_pct" = X12, 
           "tourney_finish" = X17) |>
    mutate(seed = as.numeric(seed), 
           reg_wins_pct = as.numeric(reg_wins_pct), 
           tourney_finish = as_factor(tourney_finish)) |>
    select(school, seed, reg_wins_pct, tourney_finish) 

head(basketball_data)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

```r
basketball_summary <- basketball_data |>
    rowwise() |>
    summarize(mean = mean(c_across(seed:reg_wins_pct)), 
              min = min(c_across(seed:reg_wins_pct)), 
              max = max(c_across(seed:reg_wins_pct))) |>
    drop_na()

head(basketball_summary)
```
